```python
import sys
import os

sys.path.append(os.path.abspath("../src"))
from portfolio.mean_variance import efficient_frontier
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
# -----------------------------
# Sample Data: 4 Stocks
# -----------------------------
np.random.seed(42)
returns_data = np.random.normal(0.001, 0.02, (252, 4))  # daily returns for 1 year
expected_returns = returns_data.mean(axis=0)
cov_matrix = np.cov(returns_data.T)

# -----------------------------
# Compute Efficient Frontier
# -----------------------------
returns_range = np.linspace(0.0005, 0.002, 50)  # target daily returns
ef_returns, ef_vols, ef_weights = efficient_frontier(expected_returns, cov_matrix, returns_range)

# -----------------------------
# Plot Efficient Frontier
# -----------------------------
plt.figure(figsize=(10,6))
plt.plot(ef_vols, ef_returns, 'b--', linewidth=2, label='Efficient Frontier')
plt.xlabel('Volatility (Std Dev)')
plt.ylabel('Expected Return')
plt.title('Efficient Frontier (Markowitz)')
plt.legend()
plt.grid(True)
plt.show()
```
